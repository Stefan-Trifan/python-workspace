# Ejercicio 1

lambda vs def simple  
Escribe una función que reciba un número y devuelva el doble.  
• Hazlo con def.  
• Hazlo también con lambda.  
Llama a ambas funciones con el número 5 y compara resultados.  

```python
def defDevolverDoble(num):
    return num * 2

lambdaDevolverDoble = lambda num: num * 2

print(defDevolverDoble(5))
print(lambdaDevolverDoble(5))
```

# Ejercicio 2

Dada la lista:  
nombres = ["Ana" , "Pedro" , "Luis" , "Marta"]  
Ordena la lista por la longitud del nombre usando:  
• Una función def como key.  
• Una lambda como key.  
Resultado esperado: ['Ana' , 'Luis' , 'Marta' , 'Pedro']  

```python
nombres = ["Ana", "Pedro", "Luis", "Marta"]
ordenados1 = []
ordenados2 = []

def ordenarListaDef(lista):
    return sorted(lista, key=len)

ordenarListaLambda = lambda lista: sorted(lista, key=len)

ordenados1 = ordenarListaDef(nombres)
print(ordenados1)
ordenados2 = ordenarListaLambda(nombres)
print(ordenados2)
```

# Ejercicio 3

Tienes una lista:  
nums = [1, 2, 3, 4, 5]  
• Usa map con lambda para obtener la lista de números al cubo.  
Haz otra versión usando def.  
Resultado esperado: [1, 8, 27, 64, 125]  

```python
nums = [1, 2, 3, 4, 5]

cuadrados1 = lambda x: x ** 3

def cuadrados2(x):
    return x ** 3

print(list(map(cuadrados1, nums)))
print(list(map(cuadrados2, nums)))
```

# Ejercicio 4

Define una función que reciba un número variable de argumentos (*args) y devuelva la suma de todos.  
Ejemplo:  
suma(1, 2, 3, 4) # 10  
Haz también una versión con lambda  

```python
nums = [1,2,3,4]

def sumarNumeros(*args):
    total = 0

    for x in args:
        total += x
        
    return total

print(sumarNumeros(*nums))
```

# Ejercicio 5

Define una función que reciba pares clave/valor con **kwargs y devuelva un string con el formato:  
"clave1=valor1, clave2=valor2, ..."  
Ejemplo:  
    mostrar_info(nombre="Ana" , edad=20, ciudad="Madrid")  
    # "nombre=Ana, edad=20, ciudad=Madrid"  

```python

def mostrar_info(**kwargs):

    for clave, valor in kwargs.items():
        print(clave + "=" + str(valor) + ",", end="")

mostrar_info(nombre="Ana", edad=20, ciudad="Madrid")
```

# Ejercicio 6

Define una función multiplicar(a, b, c) que multiplique 3 números.
Llama a la función pasando:  
• Una lista [2, 3, 4] usando *lista.  
• Una tupla (2, 3, 4) usando *tupla.  
• Un diccionario {"a": 2, "b": 3, "c": 4} usando **diccionario.  
Resultado esperado: 24  

```python
def multiplicar(a, b, c):
    return a * b * c

lista = [2,3,4]
tupla = (2,3,4)
diccionario = {"a":2,"b":3,"c":4}

print(multiplicar(*lista))
print(multiplicar(*tupla))
print(multiplicar(**diccionario))
```
