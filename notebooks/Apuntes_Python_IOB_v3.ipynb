# Introducción rápida a Python

> **Chuleta — Introducción a Python (por temas)**
### Tipos y utilidades
- `type(x)` — Tipo en tiempo de ejecución. Ej: `type(3.14)`.
- `is` / `is not` — Identidad de objeto. Ej: `a is b`.
- `len(x)` — Longitud. Ej: `len([1,2,3])`.
- `in` — Pertenencia. Ej: `'a' in 'casa'`.
- `range(i, j, k)` — Rangos para bucles. Ej: `range(0,10,2)`.
- `enumerate(iter)` — Índice+valor. Ej: `for i,x in enumerate(v): ...`.
- `zip(a, b)` — Iteración paralela. Ej: `for x,y in zip(A,B): ...`.
- `lambda args: expr` — Función anónima. Ej: `lambda x: x**2`.

### Cadenas y secuencias
- `s[i]` — Indexación. Ej: `'hola'[0]`.
- `s[i:j:k]` — Slicing. Ej: `lista[1:5:2]`.

### Listas
- `list.append(x)` — Añade al final.
- `list.extend(iter)` — Concatena otro iterable.
- `list.insert(i, x)` — Inserta en posición.
- `list.remove(x)` — Borra primera aparición.
- `list.pop([i])` — Extrae por índice (o último).
- `list.index(x)` — Primer índice de `x`.
- `list.count(x)` — Ocurrencias de `x`.
- `list.sort(key=..., reverse=...)` — Orden in-place.
- `list.reverse()` — Invierte in-place.

### Conjuntos
- `set(iter)` — Crear conjunto.
- `a.add(x)` / `a.discard(x)` — Añadir / Quitar sin error.
- `a | b` — Unión; `a & b` — Intersección; `a - b` — Diferencia; `a ^ b` — Dif. simétrica.
- `frozenset(iter)` — Conjunto inmutable.

### Diccionarios
- `d[k]` — Acceso (lanza `KeyError` si no existe).
- `d.get(k, def)` — Acceso con valor por defecto.
- `d.keys()` / `d.values()` / `d.items()` — Vistas claves/valores/pares.
- `d.update(otro)` — Fusiona pares clave-valor.

### Control de flujo
- `if/elif/else` — Condicionales.
- `for ... in ...` / `while ...` — Bucles.

```python
# Mini-ejemplos intro
nums = [1,2,3]
for i, x in enumerate(nums):
    print(i, x)
d = {'a':1,'b':2}
print(d.get('z', 0), d['a'])
print([x**2 for x in range(5) if x%2==0])
```

# Funciones — patrones prácticos

> **Chuleta — Funciones (por temas)**
### Definición y retorno
- `def f(args): ...` — Define función.
- `return valor` — Devuelve resultado (no imprime).
- `"""Docstring"""` — Documenta qué hace, parámetros y retorno.

### Parámetros y aridad
- `f(x, base=10)` — Parámetro por defecto.
- `def g(*args, **kwargs)` — Aridad variable.
- `g(*lista)` / `g(**dict)` — Desempaquetado al llamar.

### Errores y validación
- `try/except` — Manejo de excepciones. Ej: `except ZeroDivisionError: ...`.
- `assert cond, msg` — Chequeo durante desarrollo.

### Estilo funcional
- `map(f, xs)` — Aplica `f` a cada elemento.
- `filter(pred, xs)` — Filtra por predicado.
- `reduce(f, xs, init)` — Reduce a un acumulado.


```python
from functools import reduce

def potencia(base, exp=2):
    """Eleva 'base' a 'exp'."""
    return base**exp

def suma(*args):
    return reduce(lambda a,b:a+b, args, 0)

def seguro_div(a, b):
    try:
        return a/b
    except ZeroDivisionError:
        return float('inf')

print(potencia(3), potencia(2,10))
print(suma(1,2,3,4))
print(seguro_div(5,0))
```

# NumPy — métodos y patrones básicos

> **Chuleta — NumPy (por temas)**
### Creación de arrays
- `np.array(obj, dtype=...)` — Desde lista/iterable.
- `np.arange(inicio, fin, paso)` — Rango entero/float.
- `np.linspace(inicio, fin, num)` — `num` puntos equiespaciados.
- `np.zeros(shape)` / `np.ones(shape)` / `np.full(shape, val)` — Inicialización.

### Tipos y conversión
- `a.dtype` — Tipo del array.
- `a.astype(nuevo_dtype)` — Conversión de tipo (copia).

### Forma y reorganización
- `a.shape` — Dimensiones.
- `a.reshape(nueva_shape)` — Reorganiza.
- `a.ravel()` / `a.flatten()` — Aplanar (vista/copia).
- `a.T` — Transpuesta.

### Indexación y selección
- `a[i, j]` / `a[i:j, k:l]` — Indexación/slicing N-D.
- `a[[idxs]]` — Indexación avanzada por lista/array.
- `a[mask]` — Máscara booleana (filtrado).

### Operaciones vectorizadas
- `np.where(cond, x, y)` — Selección/recorte sin bucles.
- *Broadcasting* — Opera arrays de formas compatibles.

### Agregaciones y ejes
- `a.sum(axis=...)` / `a.mean(axis=...)` — Suma/media por eje.
- `a.min(axis=...)` / `a.max(axis=...)` — Extremos por eje.
- `a.argmin()` / `a.argmax()` — Índice del extremo global.

### Concatenación y stacking
- `np.concatenate([a,b], axis=...)` — Une por eje.
- `np.vstack([a,b])` / `np.hstack([a,b])` — Pila vertical/horizontal.
- `np.stack([a,b], axis=...)` — Nuevo eje.

### Ordenación y selección
- `np.sort(a)` — Ordena (copia).
- `a.sort()` — Ordena in-place.
- `np.argsort(a)` — Índices para ordenar.

### Aleatoriedad (generador moderno)
- `rng = np.random.default_rng(seed)` — Generador.
- `rng.integers(bajo, alto, size)` — Enteros aleatorios.
- `rng.normal(loc, scale, size)` — Gaussiana.
- `rng.choice(xs, size)` — Elección aleatoria.

### Álgebra lineal
- `A @ B` / `np.dot(A,B)` — Producto matricial.
- `np.linalg.eig(A)` — Autovalores/autovectores.
- `np.linalg.svd(A)` — SVD.


```python
import numpy as np
rng = np.random.default_rng(0)

a = np.arange(6).reshape(2,3)
b = np.ones((2,3))
print("a+b =\n", a+b)

M = rng.integers(0, 21, size=(3,4))
rec = np.where(M>10, 10, M)
print("recorte >10 -> 10:\n", rec)

row_norm = rec / rec.sum(axis=1, keepdims=True)
print("normalizado por filas:\n", row_norm)

c = np.array([10,5,7,1])
print("argsort:", np.argsort(c))
```

# Pandas — métodos y patrones básicos

### Exportación
- `df.to_csv(path, index=False)` / `df.to_excel(path, index=False)` — Exportar.


```python
import pandas as pd
df = pd.DataFrame({
    "Country": ["AR","ES","DE","ES","US"],
    "Name": ["Ana","Luis","Marta","Pedro","John"],
    "Height": [165.3, 172.1, 168.2, 180.0, 175.5],
    "Weight": [60.2, 72.5, 68.0, 82.3, 77.0],
    "Gender": ["F","M","F","M","M"],
})

df = df.assign(BMI = df["Weight"] / (df["Height"]/100)**2)
gb = df.groupby("Country").agg(peso_medio=("Weight","mean"), altura_max=("Height","max"))
pivot = pd.pivot_table(df, index="Gender", columns="Country", values="Name", aggfunc="count")
print(df.head())
print(gb)
print(pivot)
```

# Listas — métodos y patrones útiles

> **Chuleta — Listas (por temas)**
### Creación y copia
- `list(iter)` — Construir desde iterable.
- `a.copy()` / `list(a)` — Copia superficial (shallow copy).
- `a[:]` — Slicing completo (copia superficial).

### Inserción y borrado
- `a.append(x)` — Añadir al final.
- `a.extend(iter)` — Concatenar otro iterable.
- `a.insert(i, x)` — Insertar en posición.
- `a.remove(x)` — Borrar primera aparición de `x`.
- `a.pop([i])` — Extraer por índice (o último).
- `del a[i:j]` — Borrar por slicing.

### Búsqueda y conteo
- `a.index(x)` — Posición de `x` (lanza `ValueError` si no está).
- `a.count(x)` — Ocurrencias de `x`.

### Ordenación y reverso
- `a.sort(key=..., reverse=...)` — Orden in-place.
- `sorted(a, key=..., reverse=...)` — Devuelve copia ordenada.
- `a.reverse()` — Invertir in-place.
- `reversed(a)` — Iterador invertido.

### Slicing y desestructuración
- `a[i:j:k]` — Sublista por cortes y paso.
- `x, y, *rest = a` — Desempaquetado.

### Buenas prácticas
- Evitar modificar la **misma** lista mientras se itera sobre ella.
- Para modificar mientras iteras, itera sobre `a[:]` (copia) o crea una nueva lista.


```python
# Ejemplos cortos — Listas
a = [3, 1, 2, 1]
a.append(5); a.extend([8, 13])
a.insert(1, 99)
print("a:", a)
print("index(1):", a.index(1), "count(1):", a.count(1))

b = sorted(a)          # copia ordenada
a.sort(reverse=True)   # in-place descendente
print("sorted(a)->b:", b)
print("a.sort(reverse=True):", a)

# iterar sobre copia para modificar
nums = [1,2,3,4,5]
for _ in nums[:]:
    nums.pop()
print("nums vaciada con copia:", nums)
```

# Tuplas — conceptos y patrones

> **Chuleta — Tuplas (por temas)**
### Creación y acceso
- `(1, 2, 3)` / `tuple(iter)` — Crear tupla.
- `t[i]` / `t[i:j]` — Indexación y slicing.
- `len(t)` — Longitud.

### Inmutabilidad y mutabilidad interna
- *Inmutable*: no se pueden reasignar elementos (`TypeError`).
- *Mutable dentro*: si contiene objetos mutables, éstos sí cambian.

### Desempaquetado
- `x, y = t` — Desempaquetar fijo.
- `x, *rest = t` — Desempaquetado extendido.

### Usos típicos
- Claves de diccionario (si los elementos también son inmutables).
- Retorno múltiple desde funciones: `return x, y`.


```python
# Ejemplos cortos — Tuplas
t = (10, 20, 30, 40)
print("t[1:3]:", t[1:3])
x, *rest = t
print("x:", x, "rest:", rest)

# Mutabilidad interna
t2 = ([1,2], [3,4])
t2[0].append(99)
print("t2 con listas internas mutadas:", t2)
```

# Diccionarios — operaciones clave

> **Chuleta — Diccionarios (por temas)**
### Creación y acceso
- `{'k': v, ...}` / `dict(k=v, ...)` / `dict(lista_de_pares)` — Crear dict.
- `d[k]` — Acceso (lanza `KeyError` si falta).
- `d.get(k, defecto)` — Acceso con valor por defecto.
- `k in d` — Comprobar existencia de clave.

### Inserción/actualización y borrado
- `d[k] = v` — Insertar/actualizar.
- `d.update(otro_dict)` — Fusionar pares.
- `d.pop(k[, def])` — Extraer y devolver valor; opcional defecto.
- `d.setdefault(k, def)` — Devuelve `d[k]` y si no existe, asigna `def`.
- `del d[k]` — Borrar clave.
- `d.clear()` — Vaciar.

### Vistas e iteración
- `d.keys()` / `d.values()` / `d.items()` — Vistas de claves/valores/pares.
- `for k, v in d.items(): ...` — Iterar por pares.

### Dicts con contenedores
- Guardar listas/sets como valores permite acumular información por clave.
- **Ojo**: los valores mutables se comparten por referencia.


```python
# Ejemplos cortos — Diccionarios
precios = {'manzana': 1, 'pera': 2}
precios['plátano'] = 3
print("get('naranja', 0):", precios.get('naranja', 0))

# update y setdefault
precios.update({'pera': 2.5, 'uva': 4})
hist = {}
for fruta, precio in precios.items():
    hist.setdefault(fruta, []).append(precio)
print("histórico por setdefault:", hist)

# pop con valor por defecto (sin error si falta)
print("pop('naranja', -1):", precios.pop('naranja', -1))
print("items:", list(precios.items()))
```
