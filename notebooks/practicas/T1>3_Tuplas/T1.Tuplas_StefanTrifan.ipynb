# Ejercicio 3

Crear una nueva tupla a partir de otra

for x in t:  
nueva += (x**2,)  
print(nueva)  

```python
t = (10, 20, 30)
# Recorre la tupla y crea otra nueva con los valores al cuadrado

nueva = ()
for x in t:
    nueva += (x**2,)

print(nueva)
```

Pregunta: ¿por qué se crea una tupla nueva en lugar de modificar la original?

Se crea una tupla nueva porque las tuplas no pueden modificarse una vez creadas.
Cada operación con + o += genera una nueva tupla combinando las anteriores, y reasigna la variable para que apunte a la nueva.

# Ejercicio 4

```python
t = ( ("Ana" , 17), ("Luis" , 20), ("Marta" , 15) )
# Recorre la tupla y construye un diccionario {nombre: edad}
dic = {}

for nombre, edad in t:
    dic[nombre] = edad

print(dic)
```
