## Ejercicio 1

```python
# Iterar sobre la lista original y modificarla
nums = [1, 2, 3, 4, 5]
print(nums)

for x in nums:
    print("x = ", x)
    nums.pop()
    print(nums)
```

¿cuántos elementos quedan al final? ¿Por qué?
- Ninguno. Por cada iteracion del bucle for, el metodo .pop() elimina un elemento de la lista

## Ejercicio 2

```python
# Iterar sobre una copia (lista[:]) y modificar la original
nums = [1, 2, 3, 4, 5]
print(nums)

for x in nums[:]:
    nums.pop()
    print(nums)
```

¿en qué se diferencia el resultado del anterior?
- Antes ocurria un error. A medida que aumentaba x, la lista reducia su tamaño. Por lo tanto cuando se cruzaba el valor de x con el tamaño de la lista se terminaba el bucle for. Ahora lo hemos resuelto utilizando [:], el cual crea una copia de la list anums

## Ejercicio 3

```python
# Modificar elementos en vez de eliminarlos
letras = ["a", "b", "c", "d"]
print(letras)

for x in letras[:]: # recorre una copia
    letras.remove(x) # elimina el elemento concreto
    print(letras)
```

¿qué ocurre si quitas el [:] y recorres letras directamente?
