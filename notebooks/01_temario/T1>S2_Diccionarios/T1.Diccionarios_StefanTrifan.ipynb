```python
# Ejercicio 1
# Recorre el diccionario y súmale 10 a v dentro del bucle, pero sin asignar nada al diccionario. Imprime precios después.

precios = {"manzana": 1, "pera": 2, "plátano": 3}

for v in precios.values():
    v += 10
    print(v)

print(precios)
```

¿Cambia el diccionario o no cambia?

```python
# Ejercicio 2
# Recorre precios.items() y esta vez actualiza los valores con precios[k] = v * 2

precios = {"manzana": 1, "pera": 2, "plátano": 3}

for k,v in precios.items():
    precios[k] = v * 2

print(precios)
```

Comprueba si el diccionario ha cambiado
- El diccionario ha cambiado

```python
# Ejercicio 3
# Recorre el diccionario y añade un 0 al final de cada lista de notas.

bbdd = {
    "Ana": [7, 8], 
    "Luis": [5, 6], 
    "Marta": [9, 10]
}

for alumno, notas in bbdd.items():
    print(alumno, notas)
    notas.append(0)

print("\n", bbdd) 
```

```python
# Ejercicio 4
# Si la persona es menor de 18, cambia el valor a "MENOR".
# Si es mayor o igual a 18, cambia el valor a "ADULTO".

edades = {"Ana": 17, "Luis": 20, "Marta": 15, "Juan": 22}

for i in edades:
    if edades[i] < 18:
        edades[i] = "Menor"
    else:
        edades[i] = "Mayor"
    
print(edades)
```

```python
# Ejercicio 5
# Convierte cada lista de compras en el total gastado (la suma).
# Al final, compras debe quedar con números en lugar de listas.

bbdd = {
    "Ana": [10, 20, 30],
    "Luis": [5, 15],
    "Marta": [100]
}

for i, j in bbdd.items():
    print(i, j)
    bbdd[i] = sum(j)

print("\n", bbdd)
```

```python
# Ejercico 6
# Recorre productos.items() y construye otro diccionario nuevo con las mismas claves pero con el valor "CARO" si es mayor o igual a 2, o "BARATO" si es menor de 2.

productos = {
    "manzana": 1, 
    "pera": 2, 
    "plátano": 3
}

resultados = {}

for k,v in productos.items():
    if v >= 2:
        resultados[k] = "CARO" 
    else:
        resultados[k] = "BARATO"

print(resultados)
```
