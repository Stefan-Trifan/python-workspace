## B. LIMPIEZA DATAFRAME
- B0. Eliminamos las columnas que no contienen los datos previstos en el listado de campos
- B1. Trabajaremos SOLO con las columnas 1,4,5,6,8 y 9
- B2. Transformamos las columnas a los tipos de datos que nos interesan

```python
# B0. Elimina columnas, almacena el dataframe en df_limpio y comprueba que ya no existen
df_limpio=df.drop(columns=["Column10", "Column11"])
df_limpio.info()
```

```python
#B1. Haz un nuevo dataset (df_nuevo) con las columnas de trabajo 
#AYUDA: usa .iloc o .loc y crea una copia superficial
df_nuevo=df_limpio.iloc[:,[0,3,4,5,7,8]].copy()
df_nuevo
```

```python
#B2. Transforma los datos de las columnas 1,4,5,6,8, y 9 en los correctos y comprueba los tipos
# - numeros: float o int
# . fechas con datetime
# - cadenas en string
```

```python
# COLUMNAS TIPO STRING
df_nuevo['DireccionIP']=df_nuevo['DireccionIP'].fillna('').astype("string")
df_nuevo['MetodoHTTP']= df_nuevo['MetodoHTTP'].fillna('').astype("string")
df_nuevo['Ruta']=df_nuevo['Ruta'].fillna('').astype("string")
df_nuevo['EstadoHTTP']=df_nuevo['EstadoHTTP'].fillna('').astype("string")
```

```python
# COLUMNAS FECHA
df_nuevo['FechaHora'] = pd.to_datetime(df['FechaHora'], errors='coerce')
```

```python
# COLUMNAS NUMÉRICAS
df_nuevo['Bytes'] = pd.to_numeric(df['Bytes'], errors='coerce').astype('Int64')
df_nuevo.info()
```
