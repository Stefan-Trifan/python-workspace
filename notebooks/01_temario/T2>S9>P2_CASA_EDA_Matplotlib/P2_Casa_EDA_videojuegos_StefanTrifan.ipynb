```python
# Librerías
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1) Dataset de ejemplo

```python
# Dataset sintético de videojuegos
data = {
    'titulo': [
        'The Legend of Zelda', 'Super Mario Odyssey', 'FIFA 23', 'Call of Duty', 'Minecraft', 'Fortnite', 'Elden Ring', 'Animal Crossing', 'GTA V', 'God of War'
    ],
    'ventas_millones': [30, 25, 20, 22, 200, 350, 25, 40, 185, 20],
    'plataforma': ['Nintendo', 'Nintendo', 'Multiplataforma', 'Multiplataforma', 'Multiplataforma', 'Multiplataforma', 'Multiplataforma', 'Nintendo', 'Multiplataforma', 'PlayStation'],
    'puntuacion': [97, 96, 80, 82, 93, 85, 94, 90, 91, 95],
    'precio': [60, 50, 70, 70, 30, 0, 60, 50, 40, 60]
}
df = pd.DataFrame(data)
display(df)
```

#### RECORDATORIO:
En este caso podemos ver los datos del dataset facilmente (es pequeño). Recuerda que, si es dataset es muy extenso es preciso hacer antes un análisis exploratorio, como el que se hizo en P1

## 2) Histograma — Distribución de puntuaciones

```python
plt.figure()
plt.hist(df['puntuacion'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribución de puntuaciones de videojuegos')
plt.xlabel('Puntuación (Metacritic)')
plt.ylabel('Frecuencia')
plt.show()
```

**Tu interpretación:**

- ¿Cómo se distribuyen las puntuaciones?
- ¿Hay concentración en valores altos o bajos?
- ¿Alguna observación sobre la calidad general de los juegos?

> ✏️ Escribe aquí tu interpretación:

- ¿Cómo se distribuyen las puntuaciones?
    - Las puntuaciones están bastante agrupadas en la parte alta. No hay valores extremos por abajo. Todos los juegos superan con facilidad el 80. Dentro de ese rango, varios títulos se concentran cerca del 95, lo que hace que la distribución tienda a acumularse en la zona superior.

- ¿Hay concentración en valores altos o bajos?
    - Sí, hay una concentración en valores altos. La mayoría de títulos se muevem entre 90 y 97, y solo unos pocos bajan ligeramente hacia los ochenta. El rango alto domina bastante más que el medio.

- ¿Alguna observación sobre la calidad general de los juegos?
    - La calidad general del catálogo parece muy buena. Todos los juegos tienen buenas valoraciones y una parte importante son excelentes. Esto sugiere que el conjunto está formado por juegos muy bien recibidos por los jugadores.

## 3) Barras — Ventas totales por plataforma

```python
ventas_plataforma = df.groupby('plataforma')['ventas_millones'].sum()
plt.figure()
plt.bar(ventas_plataforma.index, ventas_plataforma.values, color='orange')
plt.title('Ventas totales por plataforma')
plt.xlabel('Plataforma')
plt.ylabel('Ventas (millones)')
plt.xticks(rotation=20)
plt.show()
```

**Tu interpretación:**

- ¿Qué plataforma concentra más ventas?
    - Multiplataaforma, seguida de Nintento y últimamente PlayStaiton
- ¿Qué puedes deducir sobre el tipo de juegos que más venden?
    - Los juegos multiplataforma son por diferencia los que más se venden. Al estar disponibles en varias consolas y en PC, llegan a un público muchísimo más amplio y, por tanto, acumulan muchas más ventas totales.
