```python
# Librerías
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## 1) Dataset de ejemplo

```python
# Dataset sintético de videojuegos
data = {
    'titulo': [
        'The Legend of Zelda', 'Super Mario Odyssey', 'FIFA 23', 'Call of Duty',
        'Minecraft', 'Fortnite', 'Elden Ring', 'Animal Crossing', 'GTA V', 'God of War'
    ],
    'ventas_millones': [30, 25, 20, 22, 200, 350, 25, 40, 185, 20],
    'plataforma': ['Nintendo', 'Nintendo', 'Multiplataforma', 'Multiplataforma', 'Multiplataforma',
                   'Multiplataforma', 'Multiplataforma', 'Nintendo', 'Multiplataforma', 'PlayStation'],
    'puntuacion': [97, 96, 80, 82, 93, 85, 94, 90, 91, 95],
    'precio': [60, 50, 70, 70, 30, 0, 60, 50, 40, 60]
}
df = pd.DataFrame(data)
display(df)
```

#### RECORDATORIO:
En este caso podemos ver los datos del dataset facilmente (es pequeño). Recuerda que, si es dataset es muy extenso es preciso hacer antes un análisis exploratorio, como el que se hizo en P1

## 2) Histograma — Distribución de puntuaciones

```python
plt.figure()
plt.hist(df['puntuacion'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribución de puntuaciones de videojuegos')
plt.xlabel('Puntuación (Metacritic)')
plt.ylabel('Frecuencia')
plt.show()
```

**Tu interpretación:**

- ¿Cómo se distribuyen las puntuaciones?
- ¿Hay concentración en valores altos o bajos?
- ¿Alguna observación sobre la calidad general de los juegos?

## 3) Barras — Ventas totales por plataforma

```python
ventas_plataforma = df.groupby('plataforma')['ventas_millones'].sum()
plt.figure()
plt.bar(ventas_plataforma.index, ventas_plataforma.values, color='orange')
plt.title('Ventas totales por plataforma')
plt.xlabel('Plataforma')
plt.ylabel('Ventas (millones)')
plt.xticks(rotation=20)
plt.show()
```

**Tu interpretación:**

- ¿Qué plataforma concentra más ventas?
- ¿Qué puedes deducir sobre el tipo de juegos que más venden?

## 4) Dispersión — Relación entre puntuación y ventas

```python
plt.figure()
plt.scatter(df['puntuacion'], df['ventas_millones'], color='green')
plt.title('Relación entre puntuación y ventas')
plt.xlabel('Puntuación (Metacritic)')
plt.ylabel('Ventas (millones)')
plt.show()
```

## 5) Boxplot — Distribución de precios
