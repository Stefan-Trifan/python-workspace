### 4.1) Histograma
**¿Para qué sirve?** Visualiza la **distribución** de las variables numérica y su forma (asimetrías, colas, multimodalidad).

```python
# Haz los histogramas de las columnas numéricas
df = pd.DataFrame({
    'edad': np.random.normal(35, 10, n).clip(18, 80).round(0),
    'ingresos': np.random.normal(25000, 8000, n).clip(8000, 80000).round(0),
    'altura_cm': np.random.normal(170, 10, n).clip(140, 210).round(1),
    'sexo': np.random.choice(['F', 'M'], size=n, p=[0.52, 0.48]),
    'ciudad': np.random.choice(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'], size=n)
})

plt.figure()
plt.hist(df['edad'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribución de la edad')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.show()

plt.figure()
plt.hist(df['ingresos'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribución de los ingreosos')
plt.xlabel('ingresos')
plt.ylabel('Frecuencia')
plt.show()

plt.figure()
plt.hist(df['altura_cm'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribución de las alturas')
plt.xlabel('altura')
plt.ylabel('Frecuencia')
plt.show()
```

### 4.2) Barras (frecuencias por categoría)
**¿Para qué sirve?** Comparar **frecuencias** o **totales** entre categorías.

```python
#TODO
# Haz los gráficos de barras de las columnas categóricas
df = pd.DataFrame({
    'edad': np.random.normal(35, 10, n).clip(18, 80).round(0),
    'ingresos': np.random.normal(25000, 8000, n).clip(8000, 80000).round(0),
    'altura_cm': np.random.normal(170, 10, n).clip(140, 210).round(1),
    'sexo': np.random.choice(['F', 'M'], size=n, p=[0.52, 0.48]),
    'ciudad': np.random.choice(['Madrid', 'Barcelona', 'Valencia', 'Sevilla'], size=n)
})

columnas_categoricas = df.groupby('sexo')['ingresos'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Hombres y mujeres')
plt.xlabel('Sexo')
plt.ylabel('Ingresos')
plt.xticks(rotation=20)
plt.show()

columnas_categoricas = df.groupby('sexo')['altura_cm'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Hombres y mujeres')
plt.xlabel('Sexo')
plt.ylabel('altura_cm')
plt.xticks(rotation=20)
plt.show()

columnas_categoricas = df.groupby('sexo')['edad'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Hombres y mujeres')
plt.xlabel('Sexo')
plt.ylabel('edad')
plt.xticks(rotation=20)
plt.show()

columnas_categoricas = df.groupby('ciudad')['ingresos'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Ingresos')
plt.xticks(rotation=20)
plt.show()

columnas_categoricas = df.groupby('ciudad')['altura_cm'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Ciudad')
plt.xlabel('Ciudad')
plt.ylabel('altura_cm')
plt.xticks(rotation=20)
plt.show()

columnas_categoricas = df.groupby('ciudad')['edad'].sum()
plt.figure()
plt.bar(columnas_categoricas.index, columnas_categoricas.values, color='orange')
plt.title('Ciudad')
plt.xlabel('Ciudad')
plt.ylabel('edad')
plt.xticks(rotation=20)
plt.show()
```
