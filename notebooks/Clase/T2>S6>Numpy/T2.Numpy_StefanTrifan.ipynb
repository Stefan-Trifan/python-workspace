# Ejercicio 2

Dados los arrays a = [1,2,3,4,5] y b = [10,20,30,40,50]:
- Calcula la suma de ambos arrays
- Calcula el producto elemento a elemento
- Calcula el promedio del array b

```python
a = np.array([1,2,3,4,5]) 
b = np.array([10,20,30,40,50])

print(a + b)
print(a * b)
print(b.mean())
```

# Ejercicio 4

Define dos matrices A y B de 2x2 con valores del 1 al 8.
Calcula:
- A + B
- A @ B (producto matricial)
- La transpuesta de A

```python
A = np.matrix([[1,2],[3,4]]) 
B = np.matrix([[5,6],[7,8]]) 

print(A + B)
print(A @ B)
print(A.T)
```

# Ejercicio 5

Genera un array de 10 enteros aleatorios entre 0 y 100.
Muestra:
- El valor máximo y mínimo
- La media y desviación estándar
- El índice del valor máximo

```python

rng = np.random.default_rng(42)
arr = rng.integers(0, 100, 10)
print("Array:", arr)

print("Valor máximo: ", arr.max())
print("Valor mínimo: ", arr.min())
print("Media: ", arr.mean())
print("Desviación estándar: ", arr.std())
print("Índice del valor máximo: ", np.argmax(arr))
```
