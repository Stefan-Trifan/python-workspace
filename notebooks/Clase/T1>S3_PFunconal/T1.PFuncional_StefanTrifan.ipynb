# Ejercicio 1

Tienes una lista de números:  
nums = [1, 2, 3, 4, 5]  
Usa map para obtener una lista con los cuadrados de cada número.  
Resultado esperado: [1, 4, 9, 16, 25]  

```python
nums = [1, 2, 3, 4, 5]

def cuadrado(x): 
    return x**2

list(map(cuadrado, nums))
```

# Ejercicio 2

Tienes una lista:  
nums = [10, 15, 20, 25, 30]  
Usa filter para quedarte solo con los múltiplos de 10.  
Resultado esperado: [10, 20, 30]  

```python
nums = [10, 15, 20, 25, 30]

print(list(filter(lambda x: x % 10 == 0, nums)))
```

nums = [1, 2, 3, 4, 5]

Resultado esperado: 120

# Ejercicio 4

Tienes una lista de nombres:  
nombres = ["ana" , "luis" , "pedro" , "alberto"]  
• Usa primero filter para quedarte solo con los nombres que empiecen por "a" .  
• Luego usa map para ponerlos en mayúsculas.  

Resultado esperado:  
["ANA" , "ALBERTO"]  

```python
nombres = ["ana" , "luis" , "pedro" , "alberto"]


resultado = list(
    map(lambda x: x.upper(), filter(lambda x: x.startswith("a"), nombres))
)

print(resultado)
```

# Ejercicio 5

Tienes una lista de precios:  
precios = [100, 200, 300]  
• A cada precio aplícale un IVA del 21% usando map.  
• Después usa reduce para obtener el total a pagar.  
Resultado esperado: 726.0  

```python
from functools import reduce

precios = [100, 200, 300]

# Aplicar el 21% de IVA a cada precio
precios_con_iva = map(lambda x: x * 1.21, precios)

# Calcular el total a pagar sumando todos los precios con IVA
total = reduce(lambda x, y: x + y, precios_con_iva)

print(total)
```

# Ejercicio 6

Tienes una lista de diccionarios:  
alumnos = [  
{"nombre": "Ana" , "nota": 7},  
{"nombre": "Luis" , "nota": 4},  
{"nombre": "Marta" , "nota": 9},  
{"nombre": "Pedro" , "nota": 5}  
]  
• Usa filter para quedarte solo con los que tienen nota ≥ 5.  
• Usa map para obtener solo las notas.  
• Usa reduce para calcular la media.  
Resultado esperado: 7.0 # (7 + 9 + 5) / 3  

```python
from functools import reduce

alumnos = [
    {"nombre": "Ana", "nota": 7},
    {"nombre": "Luis", "nota": 4},
    {"nombre": "Marta", "nota": 9},
    {"nombre": "Pedro", "nota": 5}
]


aprobados = list(filter(lambda alumno: alumno["nota"] >= 5, alumnos))
notas = list(map(lambda alumno: alumno["nota"], aprobados))
media = reduce(lambda x, y: x + y, notas) / len(notas)

print(aprobados)
print(notas)
print(media)
```
