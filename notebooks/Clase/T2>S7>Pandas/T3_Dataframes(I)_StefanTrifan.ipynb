### Reading files read file country.csv encoding='ISO-8859-1'

```python
import pandas as pd
df= pd.read_csv('/content/sample_data/country.csv', encoding='ISO-8859-1')
print(df.info) # Te hace un resumen cuantitativo
print(df.describe())#Con describe podremos devolver un resumen descriptivo con media, desviación, mínimo, máximo y cuartiles
```

#### Accessing by row number ( access to two fist rows)

```python
df.iloc[0:2]
#Con iloc imprimo de la fila 0 hasta la 1 ya que la segunda no esta contenida
```

#### Selecting some columns (select only 3 columns, later only 3 columns of 2 first records)

```python
df.iloc[0:2, 0:3]
#Con iloc imprimo de la fila 0 a la 1 y de la columna 0 hasta la 2
```

#### Filtering by condition. Apply a filter to a column

```python
df['Capital'] == 'Alger'
#Para filtrar elegimos una columna del df y ponemos un comparador == para comprobar si esta el valor que estamos buscando
#En este caso la Capital Alger
```

#### Add new column with default value

```python
df["NuevaColumna"] = "Hola mundo"
#Para crearnos una nueva columna es muy parecido al ej anterior solo que esta vez
#Asignamos el nombre que queremos para nuestra columna y le asignamos valores en este caso he puesto que tdos los indices tengan Hola mundo
df.head() #Mostramos las primeras filas
#print(df)
```

#### Add new column with calculated value

```python
#El valor calculado será la concatenacion de pais y capital
df['Resumen'] = df['Country'] + '-' + df['Capital']
df.head() #Mostramos las primeras filas
```

#### Remove column

```python
df.drop(columns=["NuevaColumna"], inplace=True)
#Con inplace true nos aseguramos de eliminar la columnal df original
df.head()#Mostramos las primeras filas
```

### Writing files with PANDAS. save dataframe in a csv file

```python
#A continuación con el siguiente comando me guardaré el csv
df.to_csv("country_edit.csv", sep= "|" ,index=False)
#Por defecto se guardará en la carpeta archivos
#sep será el separador que será mostrado entre las columnas
#Index=False hará que no se guarde la columna de indices
```
