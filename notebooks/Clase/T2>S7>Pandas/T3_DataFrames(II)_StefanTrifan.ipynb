#### 3. Load country codes

Load the file `../datasets/country_codes.xlsx` which contains a table with country names and their *alpha-2* and *alpha-3* codes.  
Display the table to understand its structure.

```python
df2 = pd.read_excel('/content/sample_data/country_codes.xlsx')

print(df2.info) #Mostramos información básica del DataFrame
df2.head() #Mostramos las primeras filas
```

#### 4. Set index, drop column

Use column `Alpha-2 code` as index. Drop `Alpha-3 code` since it won't be used.

```python
#Vamos a establecer el indice con set_index
df2.set_index('Alpha-2 code', inplace=True)
#inplace = True hace que modifiquemos el df original
df2.drop(columns=['Alpha-3 code'], inplace=True)

df2.head() #Mostramos las primeras filas, para verificar que se han aplicado los cambios
```

#### 5. Translate country codes in table people into full country names
Merge/join the two tables together in order to add a new column with the country names.

```python
# Con Merge juntamos los data frames
#Funciona muy parecido a un Join en sql
df = df.merge(df2[['Country name']], left_on='Country', right_index=True,
              how='left')
#df2[['Country name']]: Selececcionamos solo la columna que necesitamos
#left_on='Country': Especifica la columna del DataFrame izquierdo (df) que usaremos para hacer la unión.
#right_index=True: Le dice a pandas que la key para unir en el df2 está en el indice (Porque aquí están los códigos)
#how='left': Indica que se trata de una union de tipo left
df.head()
```

#### 6. Select the people from Germany

```python
#Imprime el dataframe con la condicion de que el nombre del pais sea igual a Germany
estudiantes_alemanes = df[df['Country name'] == 'Germany']
print(estudiantes_alemanes)
```

#### 7. Calculate average weight by country

```python
avg_by_country = df.groupby('Country name')['Weight'].mean()
print(avg_by_country.head())
#De esta manera calculamos la media del peso y la agrupamos por paises
```

#### 8. Calculate minimum and maximum values for height by country
Display minimum and maximum heights per country. Results must appear in a single table.

```python
df.groupby('Country name')['Height'].agg(['min', 'max'])
#El agg aplica ambas funciones de agregacion tanto el minimo como el maximo
```

#### 9. Calculate averages by country and gender
Let's calculate the average by country and gender at the same time.

#### 9.1. Group by country and gender together
Group by `Country name` and `Gender` in the same `groupby()`, to get aggregated values in two levels.

```python
df.groupby(['Country name', 'Gender'])[['Height','Weight']].mean()
#Agrupamos las filas en Country name y Gender y calculamos la media
#IMPORTANTE: Country name y Gender pasan a ser index
```

#### 9.2. Pivot table
Let's pivot the table to display countries in columns and gender in rows.

```python
#Antes de pivotar las tablas habrá que resetear el indice para que vuelvan a ser columnas normales
#Ya que en el ej anterior los establecimos como indices
avg_by_country_gender = df.groupby(['Country name', 'Gender'])[['Height', 'Weight']].mean().reset_index()

height_pivot = avg_by_country_gender.pivot(index='Gender', columns='Country name', values='Height')
print(height_pivot)

weight_pivot = avg_by_country_gender.pivot(index='Gender', columns='Country name', values='Weight')
print(weight_pivot)
```

#### 10. Save results
Save to Excel (or CSV) the pivot table.

```python
height_pivot.to_excel('height_pivot.xlsx')
weight_pivot.to_excel('weight_pivot.xlsx')
#Por defecto se nos guardarán en la carpeta /archivos/sample_data/
```
